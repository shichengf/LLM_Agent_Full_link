## 先看一个具体问题
Agent 做了三道题，两道成功。我们要让 Python 算出成功率。先猜下面会打印什么，再运行。方括号保存列表；每对花括号是一条记录。

```python
records = [
    {"task": "查价格", "success": True},
    {"task": "求和", "success": False},
    {"task": "查库存", "success": True},
]
for record in records:
    print(record["task"], record["success"])

success_count = sum(record["success"] for record in records)
print("总任务:", len(records))
print("成功数:", success_count)
print("成功率:", success_count / len(records))
```

## 改一个输入
`True` 在这里加起来相当于 1，`False` 相当于 0。下面增加一条成功记录。先算分子和分母；然后把新增记录改成失败，再运行本格。

```python
new_records = records + [{"task": "单位换算", "success": True}]
print("成功率:", sum(r["success"] for r in new_records) / len(new_records))
```

## 写成一个会拒绝坏输入的函数
为什么要检查类型？非空字符串 `"False"` 在布尔判断里也算真，但它不是任务成功。空列表也不能直接除以零。

```python
def success_rate(rows):
    if not rows:
        return None
    for row in rows:
        if type(row["success"]) is not bool:
            raise ValueError("success 必须是 True 或 False，不能是字符串")
    return sum(row["success"] for row in rows) / len(rows)

print(success_rate(records))
print(success_rate([]))
try:
    success_rate([{"success": "False"}])
except ValueError as error:
    print("抓到输入错误:", error)
```
