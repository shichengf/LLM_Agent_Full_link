## 字典和文件内容有什么不同？
字典是 Python 内存里的对象。JSON 是文本。JSONL 把每条记录写在独立一行。先观察转换前后的类型。

```python
import json
row = {"task": "求和", "success": True}
text = json.dumps(row, ensure_ascii=False)
print(type(row), row)
print(type(text), text)
print(json.loads(text) == row)
```

## 不用磁盘，先把管道走通
先在字符串里模拟一个两行文件。运行后手动删除第二行右花括号，看看哪个步骤失败。

```python
rows = [{"task": "A", "success": True}, {"task": "B", "success": False}]
jsonl = "\n".join(json.dumps(row) for row in rows)
print(jsonl)
restored = [json.loads(line) for line in jsonl.splitlines()]
print("还原后的条数:", len(restored))
```

## 给错误加上能定位的行号
先运行下面完整例子。再把破损行修好，确认错误消失。这里捕获异常是为了展示；真实训练数据处理应明确决定失败或跳过策略。

```python
broken = '{"success": true}\n{"success": false'
for line_number, line in enumerate(broken.splitlines(), start=1):
    try:
        print("读取成功:", json.loads(line))
    except json.JSONDecodeError as error:
        print(f"第 {line_number} 行损坏: {error.msg}")
```

## 验收与答案
Python True 在 JSON 里写成 true；两者都不是字符串。坏文件出错在第 2 行。能解释 line_number 从哪里来，即可回到讲义第 02 课，在本地真正读写 JSONL 文件。
